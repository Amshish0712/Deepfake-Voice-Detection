```python
import os
import torch
import librosa
import numpy as np
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
from transformers import HubertModel, Wav2Vec2Processor
from tqdm import tqdm
```

```python
class AudioDataset(Dataset):
    def __init__(self, folder):
        self.paths = []
        self.labels = []

        for label, cls in enumerate(["real", "fake"]):
            cls_path = os.path.join(folder, cls)
            
            for file in os.listdir(cls_path):
                if file.endswith(".wav"):
                    self.paths.append(os.path.join(cls_path, file))
                    self.labels.append(label)

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        path = self.paths[idx]
        label = self.labels[idx]

        audio, _ = librosa.load(path, sr=16000)

        # Fix length (5 sec)
        max_len = 16000 * 5
        if len(audio) > max_len:
            audio = audio[:max_len]
        else:
            audio = np.pad(audio, (0, max_len - len(audio)))

        return torch.tensor(audio, dtype=torch.float32), torch.tensor(label)
```

```python
train_data = AudioDataset(r"C:\Users\amita\Desktop\Jupyter\Dataset\training")
val_data   = AudioDataset(r"C:\Users\amita\Desktop\Jupyter\Dataset\validation")
test_data  = AudioDataset(r"C:\Users\amita\Desktop\Jupyter\Dataset\testing")

train_loader = DataLoader(train_data, batch_size=2, shuffle=True)
val_loader   = DataLoader(val_data, batch_size=2)
test_loader  = DataLoader(test_data, batch_size=2)

print("Train:", len(train_data))
print("Val:", len(val_data))
print("Test:", len(test_data))
```

```python
from transformers import HubertModel, Wav2Vec2FeatureExtractor
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

processor = Wav2Vec2FeatureExtractor.from_pretrained("facebook/hubert-base-ls960")
hubert = HubertModel.from_pretrained("facebook/hubert-base-ls960").to(device)
```

```python
def extract_features(audio):
    inputs = processor(audio, sampling_rate=16000, return_tensors="pt", padding=True)
    input_values = inputs.input_values.to(device)

    outputs = hubert(input_values)
    features = outputs.last_hidden_state  # (B, T, 768)

    # Mean pooling
    features = torch.mean(features, dim=1)

    return features
```

```python
class HubertClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(768, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 2)
        )

    def forward(self, x):
        return self.net(x)
```

```python
model = HubertClassifier().to(device)

optimizer = torch.optim.Adam(
    list(model.parameters()) + list(hubert.parameters()),  # fine-tune HuBERT
    lr=1e-5
)

criterion = nn.CrossEntropyLoss()
```

```python
def train_epoch(loader):
    model.train()
    total_loss = 0

    for audio, label in tqdm(loader):
        audio = audio.numpy()
        label = label.to(device)

        features = extract_features(audio)
        outputs = model(features)

        loss = criterion(outputs, label)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(loader)
```

```python
def evaluate(loader):
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for audio, label in loader:
            audio = audio.numpy()
            label = label.to(device)

            features = extract_features(audio)
            outputs = model(features)

            _, pred = torch.max(outputs, 1)

            total += label.size(0)
            correct += (pred == label).sum().item()

    return 100 * correct / total
```

```python
EPOCHS = 5

for epoch in range(EPOCHS):
    loss = train_epoch(train_loader)
    val_acc = evaluate(val_loader)

    print(f"\nEpoch {epoch+1}")
    print(f"Loss: {loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.2f}%")
```
