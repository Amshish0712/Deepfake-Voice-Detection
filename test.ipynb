```python
import numpy as np
import librosa
from tkinter import Tk, filedialog
from tensorflow.keras.models import load_model
```

```python
model = load_model("deepfake_voice_model_cnn.keras")   # or your saved model
```

```python

def extract_features(file_path, max_pad_len=174):
    
    try:
        audio, sr = librosa.load(file_path, sr=16000)
        
        if len(audio) == 0:
            return None
        
        mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=40)
        
        if mfcc.shape[1] < max_pad_len:
            pad_width = max_pad_len - mfcc.shape[1]
            mfcc = np.pad(mfcc, ((0,0),(0,pad_width)), mode='constant')
        else:
            mfcc = mfcc[:, :max_pad_len]
        
        return mfcc
    except:
        return None
```

```python
file_path = input("Enter full path of audio file: ").strip().replace('"','')
print("Selected file:", file_path)
```

```python

# 🔹 Process + Predict
features = extract_features(file_path)

if features is not None:
    features = features[np.newaxis, ..., np.newaxis]  # reshape
    
    prediction = model.predict(features)[0][0]
    
    label = "Fake Voice ❌" if prediction > 0.5 else "Real Voice ✅"
    
    print("\nPrediction:", label)
else:
    print("Error processing audio")
```
